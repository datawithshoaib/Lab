### Required output:
"Bob, you are doing an amazing job learning LangGraph!" 

```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

```python
# create an agent state
class AgentState(TypedDict):
    message: str

def encourage(state: AgentState) -> AgentState:
    """return an encouraging message to the user."""
    state['message'] = state['message'] + ', you are doing an amazing job learning LangGraph!'
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node(node='encourage', action=encourage)
graph.set_entry_point(key='encourage')
graph.set_finish_point(key='encourage')

app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
output = app.invoke({'message': 'Bob'})
output['message']
```
